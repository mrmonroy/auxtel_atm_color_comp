```python
from astroquery.simbad import Simbad
```

```python
Simbad.add_votable_fields('otype')      #(before 'maintype')
Simbad.add_votable_fields('sp_type')    #(before 'sptype')

ra_min, ra_max = 149.5, 150.7
dec_min, dec_max = 1.7, 2.7
vmag_min, vmag_max = 18, 20

query = f"""
    SELECT basic.main_id, basic.ra, basic.dec, flux.V, basic.otype, basic.sp_type
    FROM basic
    JOIN allfluxes AS flux ON basic.oid = flux.oidref
    WHERE
        basic.ra BETWEEN {ra_min} AND {ra_max} AND
        basic.dec BETWEEN {dec_min} AND {dec_max} AND
        flux.V BETWEEN {vmag_min} AND {vmag_max} AND
        basic.otype = 'Star'
"""

result = Simbad.query_tap(query)

if result is not None:
    print(f"{len(result)} 'Star' objects found:")
    print(result)
    # Guardar en CSV
    result.write('results_stars.csv', format='csv', overwrite=True)
else:
    print("No Star objects found in the specified ranges")
```
