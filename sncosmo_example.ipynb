```python
import numpy as np
import sncosmo

wave = np.arange(3000, 9000, 10)
model = sncosmo.Model(source='hsiao-subsampled')
model.set(z=0.1, amplitude=1e-5, t0=0.)
sample_time = model['t0'] + 2.
flux = model.flux(time=sample_time, wave=wave)
fluxerr = 0.1 * flux
flux = flux + np.random.normal(scale=fluxerr)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(wave,flux)
```

```python
wave = np.arange(3000, 9000, 10)
model = sncosmo.Model(source='hsiao-subsampled')
model.set(z=0.5, amplitude=1e-5, t0=1.)
sample_time = model['t0'] + 2.
flux = model.flux(time=sample_time, wave=wave)
fluxerr = 0.1 * flux
flux = flux + np.random.normal(scale=fluxerr)
```

```python
plt.plot(wave,flux)
```

```python
wave = np.arange(3000, 9000, 10)
model = sncosmo.Model(source='hsiao-subsampled')
for z_ in np.linspace(0.,1.0,10):
    model.set(z=z_, amplitude=1e-5, t0=0.)
    sample_time = model['t0'] + 2.
    flux = model.flux(time=sample_time, wave=wave)
    plt.plot(wave,flux)
```
